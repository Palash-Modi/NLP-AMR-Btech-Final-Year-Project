```python
import torch

# Load the file
data = torch.load('best.pt', map_location=torch.device('cpu'))

print("File object type:", type(data))

# Safely inspect the dictionary contents
if isinstance(data, dict):
    for key, value in data.items():
        if hasattr(value, 'shape'):
            print(f"Tensor Key: {key} | Shape: {value.shape} | Type: {value.dtype}")
        elif isinstance(value, dict):
            print(f"Dict Key:   {key} | Sub-keys: {list(value.keys())[:5]}...")
        else:
            print(f"Other Key:  {key} | Type: {type(value).__name__} | Value: {value}")
else:
    print("Data is not a dictionary. Direct value:", data)
```

```python
import torch

# Load the file
data = torch.load('best.pt', map_location=torch.device('cpu'))

# Extract the model object
model = data['model']

# Get the classes dictionary (usually mapping integer IDs to string names)
if hasattr(model, 'names'):
    classes = model.names
    print(f"🎉 Found {len(classes)} classes:\n")
    for idx, name in sorted(classes.items()):
        print(f"Class {idx}: {name}")
else:
    print("Could not find the '.names' attribute on the model object.")
```
